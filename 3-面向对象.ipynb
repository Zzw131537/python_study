类: 用来描述具有相同属性和方法的对象的集合
方法: 类中定义的函数
数据成员: 类变量或实例变量
方法重写: 如果从父类继承的方法不能满足子类的需求，可以对其进行改写
继承: 一个派生类继承基类的字段和方法
实例化: 创建一个类的实例
对象: 通过类定义的数据实例


```python
from datetime import datetime, timedelta


class MyClass:
    def __init__(self,value):
        self.value = value

    def display_value(self):
        print(self.value)

obj = MyClass(5)
obj.display_value()

# 继承
class People:
    name=''
    age=''
    _weight=0

    def __init__(self,n,a,w):
        self.name=n
        self.age=a
        self._weight=w
    def speak(self):
        print("%s 说: 我 %d岁" %(self.name,self.age))

class Student(People):
    grade=''

    def __init__(self,n,a,w,g):
        People.__init__(self,n,a,w)
        self.grade=g

    #方法重写
    def speak(self):
        print("%s 说: 我 %d 岁了，我在读 %d 年级。" %(self.name,self.age,self.grade))

s=Student('bob',10,60,3)
s.speak()
```

创建一个类来实现数字的加减乘除运算

```python
class Calculator:
    def __init__(self,num1,num2:int):
        self.num1=num1
        self.num2=num2

    def add(self):
        return self.num1+self.num2

    def subtract(self):
        return self.num1-self.num2

    def multiply(self):
        return self.num1*self.num2

    def divide(self):
        if self.num2 == 0:
            return "Error: Division by zero is not allowed"
        return self.num1/self.num2

calc = Calculator(9,18)
print(calc.add())
print(calc.subtract())
print(calc.multiply())
print(calc.divide())
```

实现一个类，支持日期的加减操作

```python
from datetime import datetime,timedelta
class Date:
    def __init__(self,year,month,day):
        self.date = datetime(year,month,day)

    def __add__(self, days):
        new_date = self.date + timedelta(days=days)
        return Date(new_date.year,new_date.month,new_date.day)

    def __sub__(self, days):
        new_date = self.date - timedelta(days=days)
        return Date(new_date.year,new_date.month,new_date.day)

    def __str__(self):
        return self.date.strftime("%Y-%m-%d")


date=Date(2026,5,1)
print(date)

data_2=date+2
print(data_2)

date_3=date-3
print(date_3)
```
