### 字符串:
- 单引号和双引号的使用完全相同
- 使用三引号可以指定一个多行字符串
- 转义符使用 '\'
- '+' 可以连接字符串,'*'可以重复字符串
- 字符串不能改变
- str[start:end:step]: 其中start(包含)是切边的开始,end(不包含)是切边的结束，step是切片的步长

```python
s = '123456789'

print(s)
print(s[0:-1])
print(s[0])
print(s[2:5])
print(s[2:])
print(s[1:5:2])
print(s*2)
print(s+'zzw')

print('-'*20)
print('hello\n word')
print(r'hello\n world') #字符串前加 r ，表示原始字符串，不会发生转义
```

### 标准数据类型
Python 中有6中标准数据类型(bool类型是int的子类,有时单独列出)
- Number(数字)
- String(字符串)
- bool(布尔类型)
- List(列表)
- Tuple(元组)
- Set(集合)
- Dictionary(字典)

按是否可变，分成以下两类:
- 不可数据(4个): Number,String,bool,Tuple
- 可变类型(3个): List,Dictionary,Set


```python
num1=1
num2=2
print(num1+num2)
print(num1-num2)
print(num1*num2)
print(num1/num2) # 除法,得到一个浮点数
print(num1//num2) # 整除,得到一个整数
print(num1%num2) # 取余
print(num1**num2) # 乘方
```

bool类型只有两个值: True(1),False(0)
只有 0 ,空字符串，空列表，空元组等被视为 False,其他的均视为 True

```python
print(True)
print(False)
print(True==1)
print(False==0)
```

列表中类型可以并不相同，和字符串一样，列表同样可以被索引和截取,被截取后返回一个包含截取元素的新列表,但列表中的元素是可以被改变的
变量[头下标:尾下标]

```python
my_list=['abcd',786,2.23,True]
print(my_list)
print(my_list[0])
print(my_list[1:3])
print(my_list[2:])
print(my_list*2)
my_list[0]='123'
print(my_list)
```

元组与列表类似，不同之处在于元组中的元素不能改变,可以将字符串看成一串特殊的元组

```python
my_tuple=('abcd',123,2.23)
print(my_tuple)
print(my_tuple[0])
print(my_tuple[1:2])
print(my_tuple[1:])
print(my_tuple*2)
```

集合用于存储唯一的元素，可以进行交，并等常见的集合操作

```python
set1={'a','b','c'}
set2={'c','d','e'}
print(set1)

if 'a' in set1:
    print('Y')
else:
    print('N')

print(set1 - set2)
print(set1 | set2)
print(set1 & set2)
print(set1 ^ set2)
```

字典是一种映射类型, 它是一个键值对,其中 K必须是不可变类型,且一个字典中的 K 是唯一的

```python
my_dict ={}
my_dict['a'] = '1'
my_dict['b']='2'
print(my_dict['a'])
print(my_dict)
print(my_dict.keys())
print(my_dict.values())
print(my_dict.items())
```

隐式类型转换: 会自动将一种数据类型转换成另一种数据类型

```python
num1=123
num2=1.23
num3=num1+num2
print(type(num1))
print(type(num2))
print(type(num3))
```

条件控制: 'if','elif','else'

```python
num1=3
if num1 > 3:
    print('>')
elif num1 < 3:
    print('<')
else:
    print('=')
```

条件控制: match...case

```python
num1=1
match num1:
    case 2: # elif num1 == 2
        print('2')
    case 3: # elif num1 == 3
        print('3')
    case _: # else
        print(num1)
```
