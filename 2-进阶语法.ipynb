Python 推导式: 从一个数据序列构建出另一个新的数据序列。

```python
from time import process_time

my_list = ['123','1234']
a = [x for x in my_list if len(x) > 3]
b = [x for x in range(30) if x % 3 == 0]

my_dict = {x: x**2 for x in b}
print(a)
print(b)
print(my_dict)
```

迭代器是方位集合元素的一种方式。字符串，列表，元组对象都可以创建迭代器

```python
my_list=[1,2,3,4]
it = iter(my_list)
print(next(it))
print(next(it))
for x in it:
    print(x)

# 创建一个数字迭代器
class MyNumbers:
    def __iter__(self):
        self.a = 1
        return self

    def __next__(self):
        if self.a <= 3:
            x=self.a
            self.a += 1
            return x
        else:
            raise StopIteration

c=MyNumbers()
it=iter(c)

for x in it:
    print(x)
```

with 关键字提供了一种优雅的方式来处理文件操作，数据库连接等需要明确释放资源的场景。
with 语句通过上下文管理协议可以:
- 自动资源释放
- 代码简洁
- 异常安全
- 可读性强

```python
with open('example.txt','r') as f:
    content = f.read()
    print(content)
# 文件自动关闭
```

函数是组织好的，可重复使用的代码段。
参数传递:
不可变类型: 传递的只是变量的值，没有影响a对象本身。如果在 func(a)内部修改 a的值，则是新生成一个a的对象。
可变类型: 会将a真正的传进去，修改后func外部的a也会受影响。

```python
def max_num(a,b: int)->int:
    if a > b:
        return a
    else:
        return b

a=1
b=2
print(max_num(a,b))

def change(a:int):
    print(id(a))
    a=10
    print(id(a))

a=1
print(id(a))
change(a)

def change2(d:list):
    d.append(2)
    print(d)
    return

c=[1]
change2(c)
print(c)
```

装饰器 = 函数的包装函数 + 不修改原代码的扩展功能。python 内置了一些装饰器，这里不再赘述。

```python
def my_decorator(func):
    def wrapper():
        print('start')
        func()
        print('end')
    return wrapper

@my_decorator
def say_hello():
    print('hello')
say_hello()
```
